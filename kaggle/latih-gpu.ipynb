# Rategoan — Latih GPU (Kaggle, API resmi)

Port dari `colab/latih-gpu.ipynb`. **Token tidak ditulis di notebook.**

Secret Kaggle (Add-ons > Secrets), nama persis:
- `GITHUB_TOKEN` — PAT repo, jangan commit

Uji kecil dulu: `MODEL_SIZE=50m`, `MAX_MINUTES=5`.
Baru 200m / 60 menit setelah uji lolos.

Jalur resmi: [Kaggle Kernels API](https://github.com/Kaggle/kaggle-api/blob/main/docs/kernels.md) `kaggle kernels push`.
Colab *gratis* tidak punya API Run All. Colab Enterprise butuh proyek GCP.


```python
import os, subprocess
print(subprocess.run(['nvidia-smi'], capture_output=True, text=True).stdout)
import torch
print('torch', torch.__version__, 'cuda', torch.cuda.is_available())
if not torch.cuda.is_available():
    raise SystemExit('GPU Kaggle belum aktif. Di kernel-metadata.json set enable_gpu true, atau pilih GPU saat push.')
```

```python
# Uji kecil: 50m + 5 menit. Production: 200m + 60.
MODEL_SIZE = os.environ.get('RAGET_MODEL_SIZE', '50m')
MAX_MINUTES = int(os.environ.get('RAGET_MAX_MINUTES', '5'))
print('MODEL_SIZE', MODEL_SIZE, 'MAX_MINUTES', MAX_MINUTES)
```

```python
from kaggle_secrets import UserSecretsClient
GITHUB_TOKEN = UserSecretsClient().get_secret('GITHUB_TOKEN')
if not GITHUB_TOKEN:
    raise SystemExit('Tambahkan secret Kaggle bernama GITHUB_TOKEN. Jangan tempel token di sel.')
os.environ['GITHUB_TOKEN'] = GITHUB_TOKEN
print('Token OK, panjang', len(GITHUB_TOKEN), '(nilai tidak dicetak)')
```

```python
import subprocess, os
OWNER, REPO = 'maetalizer-png', 'Rategoan'
url = 'https://x-access-token:' + GITHUB_TOKEN + '@github.com/' + OWNER + '/' + REPO + '.git'
if os.path.isdir('Rategoan/.git'):
    subprocess.check_call(['git','-C','Rategoan','pull','--ff-only'])
else:
    subprocess.check_call(['git','clone','--depth','1', url, 'Rategoan'])
os.chdir('Rategoan')
print('repo OK')
```
